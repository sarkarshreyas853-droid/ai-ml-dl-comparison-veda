# AI vs Machine Learning vs Deep Learning

This notebook explains the difference between **Artificial Intelligence (AI)**, **Machine Learning (ML)** and **Deep Learning (DL)**, and shows a small ML example you can run.

**Tools:** Python, Jupyter Notebook, scikit-learn, Markdown.

## 1. Definitions

- **Artificial Intelligence (AI)**: the broad field of building machines that perform tasks that normally need human intelligence.
- **Machine Learning (ML)**: a subset of AI in which a system learns patterns from data instead of following hand-written rules.
- **Deep Learning (DL)**: a subset of ML that uses neural networks with many layers to learn complex patterns.

## 2. Comparison

| | AI | Machine Learning | Deep Learning |
|---|---|---|---|
| Scope | Broadest field | Subset of AI | Subset of ML |
| How it works | Rules, search, learning, or a mix | Learns patterns from data using algorithms such as decision trees or linear models | Learns features automatically with multi-layer neural networks |
| Data needs | Varies | Moderate | Large amounts of data |
| Compute needs | Varies | Usually modest | Usually high (GPUs) |
| Feature engineering | Varies | Often done by hand | Largely automatic |
| Interpretability | Varies | Often easier to explain | Harder to explain |

## 3. Real-world examples

- **AI:** a chess engine that plans moves, or a customer-support chatbot that routes questions.
- **Machine Learning:** a spam filter that learns from labelled emails, or a model that predicts house prices from past sales.
- **Deep Learning:** image recognition that tags faces in photos, or speech recognition in a voice assistant.

## 4. A small Machine Learning example

Instead of writing rules by hand, we give a model labelled examples and let it learn the rule. Here a classifier learns to tell apart two flower species from their measurements (the Iris dataset).

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

X, y = load_iris(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

model = DecisionTreeClassifier(max_depth=3, random_state=42)
model.fit(X_train, y_train)  # the model learns the patterns from data

predictions = model.predict(X_test)
print(f"Test accuracy: {accuracy_score(y_test, predictions):.2%}")
```

## 5. Deep Learning in one paragraph

A deep learning model stacks many layers of artificial neurons. Each layer turns its input into a more abstract representation, so the network learns features (edges, shapes, words) by itself. This needs much more data and compute than the decision tree above, which is why it is used for images, audio and text. Frameworks such as PyTorch or TensorFlow are normally used for it.

## 6. How they are related

Think of them as nested circles. **AI** is the largest circle. **Machine Learning** sits inside AI, because it is one of the main ways to build AI systems. **Deep Learning** sits inside Machine Learning, because it is one family of ML methods that relies on neural networks.

Recommendation systems touch all three: the product is AI, a model that learns user preferences from past behaviour is ML, and a deep neural network that learns from raw images or text to match products is DL.

## 7. Summary

- AI is the goal: machines that act intelligently.
- ML is a way to reach that goal by learning from data.
- DL is a specific kind of ML using deep neural networks, strongest on large unstructured data.
